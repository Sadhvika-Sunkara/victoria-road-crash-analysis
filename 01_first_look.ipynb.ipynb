```python
# Import the pandas library
import pandas as pd

# Full path to data files
data_path = r'C:\Users\sunka\Documents\victoria-road-crash-analysis\data\raw'

# Load all 4 CSV files
accident = pd.read_csv(f'{data_path}\\accident.csv')
person   = pd.read_csv(f'{data_path}\\person.csv')
vehicle  = pd.read_csv(f'{data_path}\\vehicle.csv')
node     = pd.read_csv(f'{data_path}\\node.csv')

# Show how big each table is
print("=== TABLE SIZES ===")
for name, df in [('accident', accident), ('person', person), 
                 ('vehicle', vehicle), ('node', node)]:
    print(f"{name}: {df.shape[0]:,} rows and {df.shape[1]} columns")
```

```python
# Look at the first 5 rows of the accident table
print("=== FIRST 5 ROWS OF ACCIDENT TABLE ===")
print(accident.head())

# What are the column names?
print("\n=== COLUMN NAMES ===")
print(accident.columns.tolist())
```

```python
# How many missing values in each column?
print("=== MISSING VALUES IN ACCIDENT TABLE ===")
print(accident.isnull().sum())

print("\n=== DATA TYPES OF EACH COLUMN ===")
print(accident.dtypes)
```

```python
# Check missing values across all 4 tables
print("=== MISSING VALUES SUMMARY ===")
for name, df in [('accident', accident), ('person', person),
                 ('vehicle', vehicle), ('node', node)]:
    total_missing = df.isnull().sum().sum()
    print(f"{name}: {total_missing:,} total missing values")
```

```python
# Key statistics about crashes
print("=== KEY FACTS ===")
print(f"Total crashes recorded: {len(accident):,}")
print(f"Total people involved: {len(person):,}")
print(f"Total fatalities: {accident['NO_PERSONS_KILLED'].sum():,}")
print(f"Date range: {accident['ACCIDENT_DATE'].min()} to {accident['ACCIDENT_DATE'].max()}")
print(f"\nCrashes by severity:")
print(accident['SEVERITY'].value_counts())
```
